```python
from app.datasets.solomon_loader import load_solomon

df = load_solomon("/app/datasets/solomon/C101.txt")

df.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,10))

plt.scatter(df["x"], df["y"])

for _, row in df.iterrows():
    plt.text(
        row["x"],
        row["y"],
        str(int(row["id"])),
        fontsize=8
    )

plt.title("Solomon Dataset C101")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)

plt.show()
```
